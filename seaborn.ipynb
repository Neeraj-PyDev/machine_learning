```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer_data = load_breast_cancer()
```

```python
cancer_data
```

```python
type(cancer_data)
```

```python
cancer_df = pd.DataFrame(np.c_[cancer_data['data'],cancer_data['target']],columns=np.append(cancer_data['feature_names'],['target']))
```

```python
cancer_df.columns
```

```python
sns.pairplot(cancer_df, vars=['mean radius', 'mean texture', 'mean perimeter', 'mean area',
       'mean smoothness'], hue= 'target')
```

```python
sns.pairplot(cancer_df, hue= 'target', x_vars = ['mean radius', 'mean texture'] , y_vars= ['mean radius'] , kind = 'reg')
```
