##Using linear regression - supervised machine learning algorithm. 

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge, Lasso
```

```python
path = r"https://drive.google.com/uc?export=download&id=1xxDtrZKfuWQfl-6KA9XEd_eatitNPnkB" 
df = pd.read_csv(path)
```

```python
pd.set_option("display.max_columns",None)
pd.set_option("display.max_rows",None)
```

```python
df.head()
```

### split data

```python
X = df.drop('price', axis=1)
y = df['price']

print('shape of X:',X.shape)
print('shape of y:',y.shape)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y, train_size=0.8, random_state=51)
```

```python
# X_train = X_train.fillna(X_train.mean())
```

```python
print('shape of X_train',X_train.shape)
print('shape of X_test',X_test.shape)
print('shape of y_train',y_train.shape)
print('shape of y_test',y_test.shape)
```

### Feature Scaling

```python
sc = StandardScaler()
sc.fit(X_train)
X_train = sc.transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

### Linear Regression- ML model Training

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
lr.coef_
```

```python
lr.intercept_
```

### Predict the value of home and test

```python
X_test[0 , :]
```

```python
lr.predict([X_test[0 , :]])
```

```python
lr.predict([X_test[301 , :]])
```

```python
lr.predict(X_test)
```

```python
y_test
```

## to check the percentage of accuracy , our model is predicting the prices of houses, we need to check score.

```python
lr.score(X_test, y_test)
```

```python
# 79 % is really a great predicting percentage but if we need to train more accurately then we can use hyperparameter etc...
```

### Ridge & Lasso 

```python
# Used for penalize the overfit model under linear regression. and train model more accurately.
```

```python
rd = Ridge()

rd.fit(X_train, y_train)
```

```python
rd.score(X_test, y_test)
```

```python
las = Lasso()

las.fit(X_train, y_train)
```

```python
las.score(X_test, y_test)
```

```python
# for more best percentage, use parameter alpha i ridge or lasso.
```

```python
las2 = Lasso(alpha=2)
las2.fit(X_train, y_train)
las2.score(X_test, y_test)
```

```python
rd2= Ridge()
rd2.fit(X_train,y_train)
rd2.score(X_test,y_test)
```
