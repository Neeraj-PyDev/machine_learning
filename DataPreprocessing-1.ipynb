```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(df.isnull())
```

```python
# calculate null val percentage
null_var = df.isnull().sum()/df.shape[0]*100
null_var
```

```python
# select columns to be dropped {1st method}


drp_col = null_var[null_var > 17].keys()
drp_col
```

```python
df1 = df.drop(columns=drp_col)
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(df1.isnull())
```

```python
# dropna {2nd method}

df2 = df1.dropna()
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(df2.isnull())
```

```python
cat_var = df2.select_dtypes(include=['object']).columns
cat_var
```

```python
pd.concat([df['MSZoning'].value_counts()/df.shape[0]*100, df2['MSZoning'].value_counts()/df2.shape[0]*100] , axis=1, keys=['org','cleaned'])
```

```python
# dynamical prog

def categorical_var_distribution(var):
    
    return pd.concat([df[var].value_counts()/df.shape[0]*100, df2[var].value_counts()/df2.shape[0]*100] , axis=1, keys=[var+'_org',var+'_cleaned'])
    
```

```python
categorical_var_distribution('MSZoning')
```

```python

def categorical_var_distribution2():
    all=[]
    for i in cat_var:
        d = pd.concat([df[i].value_counts()/df.shape[0]*100, df2[i].value_counts()/df2.shape[0]*100] , axis=1, keys=[i+'_org',i+'_cleaned'])
        all.append(d)
    return all
```

```python
categorical_var_distribution2()
```
