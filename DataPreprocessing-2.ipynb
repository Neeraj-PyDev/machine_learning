```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\neeraj\train.csv")
```

```python
df.shape
```

```python
pd.set_option("display.max_columns",None)
pd.set_option("display.max_rows",None)
```

```python
df.tail()
```

```python
null_var = df.isnull().sum()/df.shape[0]*100
drp_col = null_var[null_var > 60].keys()
drp_col
```

```python
df_1 = df.drop(columns = drp_col)
```

```python
df_1.shape
```

```python
lis = [var for var in df_1.columns if df_1[var].isnull().sum()>0]
lis
```

```python
num_var = df_1.select_dtypes(include=['int64','float64'])
num_var
```

```python
miss_in_num_var = [col for col in num_var.columns if num_var[col].isnull().sum()>0]
miss_in_num_var
```

```python
plt.figure(figsize=(10,10))
for i,var in enumerate(miss_in_num_var):
    plt.subplot(2,2,i+1)
    sns.distplot(num_var[var],bins=20)
```

```python
df_2_mean = num_var.fillna(num_var.mean())
```

```python
df_2_mean.isnull().sum().sum()
```

```python
plt.figure(figsize=(10,10))
for i,var in enumerate(miss_in_num_var):
    plt.subplot(2,2,i+1)
    sns.distplot(num_var[var], bins=20, kde_kws={'linewidth':5,'color':'red'},label='original')
    sns.distplot(df_2_mean[var], bins=20, kde_kws={'linewidth':5,'color':'green'}, label='cleaned')
    plt.legend()
```

```python
df_3_median = num_var.fillna(num_var.median())
df_3_median.isnull().sum().sum()
```

```python

for i, var in enumerate(miss_in_num_var):
    plt.figure(figsize=(10,10))
    plt.subplot(3,1,1)
    sns.boxplot(num_var[var])
    plt.subplot(3,1,2)
    sns.boxplot(df_2_mean[var])
    plt.subplot(3,1,3)
    sns.boxplot(df_3_median[var])
    
```

```python
df_1_copy = df_1.copy()
num_var_miss = ['LotFrontage', 'MasVnrArea', 'GarageYrBlt']
cat_vars_ref  = ['LotConfig', 'MasVnrType', 'GarageType'] 
for cat_vars,num_var in zip(cat_vars_ref,num_var_miss):
    for var_cls in df_1[cat_vars_ref]:
        df_1_copy.update(df_1[df_1.loc[:,cat_vars]==var_cls][num_var].replace(np.nan,df_1[df_1.loc[:,cat_vars]==var_cls][num_var].mean()))

df_1_copy.isnull().sum()

```
