```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
train = pd.read_csv("C:/neeraj/train.csv")
test = pd.read_csv("C:/neeraj/test.csv")
```

```python
X_train = train.drop(columns="SalePrice",axis=1)
y_train = train["SalePrice"]
X_test = test.copy()

print("shape of X_train:",X_train.shape)
print("shape of y_train:",y_train.shape)
print("shape of X_test:",X_test.shape)
```

```python
null_sum = X_train.isnull().sum()
```

```python
# find num var which has null

num_var = X_train.select_dtypes(include=["int64","float64"]).columns
num_nulls = [i for i in num_var if null_sum[i]>0]
num_nulls
```

```python
# find cat var which has null

cat_vars = X_train.select_dtypes(include=["object"]).columns
cat_nulls = [j for j in cat_vars if null_sum[j]>0]
cat_nulls
```

```python
num_var_mean = ['LotFrontage']
num_var_median = ['MasVnrArea', 'GarageYrBlt']
cat_var_mode = ['Alley', 'MasVnrType', 'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2', 'Electrical', 'FireplaceQu']
cat_var_constant = ['GarageType', 'GarageFinish', 'GarageQual', 'GarageCond', 'PoolQC', 'Fence', 'MiscFeature']
```

```python
num_var_mean_imputer = Pipeline(steps=[('imputer',SimpleImputer(strategy='mean'))])
num_var_median_imputer = Pipeline(steps=[('imputer',SimpleImputer(strategy='median'))])
num_var_mode_imputer = Pipeline(steps=[('imputer',SimpleImputer(strategy='most_frequent'))])
num_var_constant_imputer = Pipeline(steps=[('imputer',SimpleImputer(strategy='constant', fill_value='missingg'))])
```

```python
preprocessor = ColumnTransformer(transformers=[('mean_imputer',num_var_mean_imputer,num_var_mean),
                                               ('median_imputer',num_var_median_imputer,num_var_median),
                                               ('mode_imputer',num_var_mode_imputer,cat_var_mode),
                                               ('constant_imputer',num_var_constant_imputer,cat_var_constant)                                              
                                              ])
```

```python
X_train_clean = preprocessor.transform(X_train)
X_test_clean = preprocessor.transform(X_test)
```

```python
#numerical nulls
null_sum = X_train.isnull().sum()
num_var = X_train.select_dtypes(include=['int64','float64']).columns
num_nulls = [i for i in num_var if null_sum[i]>0]
num_nulls
```

```python
#Using ColumnTransformers and Pipeline method.

num_var_mean = ['LotFrontage']
num_var_median = ['MasVnrArea', 'GarageYrBlt']
cat_var_mode = ['Alley',
 'MasVnrType',
 'BsmtQual',
 'BsmtCond',
 'BsmtExposure',
 'BsmtFinType1',
 'BsmtFinType2',
 'Electrical',
 'FireplaceQu']
cat_var_const = ['GarageType',
 'GarageFinish',
 'GarageQual',
 'GarageCond',
 'PoolQC',
 'Fence',
 'MiscFeature']

#Create Pipeline imputers
num_var_mean_imputer = Pipeline(steps=[('imputer',SimpleImputer(strategy='mean'))])
num_var_median_imputer = Pipeline(steps=[('imputer',SimpleImputer(strategy='median'))])
cat_var_mode_imputer = Pipeline(steps=[('imputer',SimpleImputer(strategy='most_frequent'))])
cat_var_const_imputer = Pipeline(steps=[('impuetr',SimpleImputer(strategy='constant', fill_value='noavailable'))])

## create Transformers
preprocessor_2 = ColumnTransformer(transformers=[('mean_imputer' ,num_var_mean_imputer,num_var_mean),
                                                ('median_imputer' ,num_var_median_imputer,num_var_median),
                                                ('mode_imputer' ,cat_var_mode_imputer,cat_var_mode),
                                                ('const_imputer' ,cat_var_const_imputer,cat_var_const)])
```

```python
#First fit before transform

preprocessor_2.fit(X_train)
preprocessor_2.fit(X_test)

# now preprocessor_1 will change the actual data.

X_train_cleaned = preprocessor_2.transform(X_train)
X_test_cleaned = preprocessor_2.transform(X_test)
```
