```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler, PolynomialFeatures
from sklearn.metrics import mean_squared_error
```

```python
path = r"https://drive.google.com/uc?export=download&id=1xxDtrZKfuWQfl-6KA9XEd_eatitNPnkB"
df = pd.read_csv(path)
df.shape
```

```python
X = df.drop('price',axis=1)
y = df['price']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X, y, train_size=0.8, random_state=51)
```

```python
X_train.shape
```

```python
sc = StandardScaler()
sc.fit(X_train)
X_train = sc.transform(X_train)
X_test = sc.transform(X_test)
```

## Polynomial Regression

```python
poly_reg = PolynomialFeatures(degree=2)
poly_reg.fit(X_train)
X_train_poly = poly_reg.transform(X_train)
X_test_poly = poly_reg.transform(X_test)
```

```python
X_train_poly.shape , X_test_poly.shape
```

```python
# The polynomial trained model when passed to fit into LinReg , it always takes bit of time like 2-3 mins.
```

```python
lr = LinearRegression()
lr.fit(X_train_poly, y_train)
lr.score(X_test_poly, y_test)
```

```python
lr.predict([X_test_poly[0, :]])
```

```python
y_pred = lr.predict(X_test_poly)
```

```python
y_test
```

```python
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print('mse:',mse ,',','rmse:',rmse)
```
